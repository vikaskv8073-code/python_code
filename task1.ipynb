```python
print("Using a FOR loop:")

for i in range(1, 101):
    print(i, end=" ")  
print("\n")

print("Using a WHILE loop:")
j = 1
while j <= 100:
    print(j, end=" ")
    j += 1
```

```python
num = int(input("Enter a number to find its factorial: "))
factorial = 1

if num < 0:
    print("Factorial does not exist for negative numbers.")
elif num == 0:
    print("The factorial of 0 is 1")
else:
    for i in range(1, num + 1):
        factorial = factorial * i
    print(f"The factorial of {num} is {factorial}")
```

```python
my_list = [45, 12, 89, 33, 7, 21, 99, 54, 60, 2]
print(f"Original list: {my_list}")

my_list.append(100)
print(f"After append(100): {my_list}")

if 33 in my_list:
    my_list.remove(33)
print(f"After remove(33): {my_list}")

my_list.sort()
print(f"After sort(): {my_list}")

my_list.reverse()
print(f"After reverse(): {my_list}")
```

```python
text = input("Enter a string: ")
vowels = "aeiouAEIOU"
count = 0

for char in text:
    if char in vowels:
        count += 1

print(f"Number of vowels in the string: {count}")
```

```python
word = "radar"
reversed_word = word[::-1]

if word == reversed_word:
    print(word, "is a palindrome!")
else:
    print(word, "is NOT a palindrome.")
```

```python
def is_palindrome(text):
    cleaned_text = str(text).replace(" ", "").lower()
    return cleaned_text == cleaned_text[::-1]

user_string = input("Enter a string to check: ")

if is_palindrome(user_string):
    print(f"'{user_string}' is a palindrome!")
else:
    print(f"'{user_string}' is not a palindrome.")
```
